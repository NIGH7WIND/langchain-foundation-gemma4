## Initialising and invoking an agent

```python
from langchain.agents import create_agent

gemma_model = init_chat_model(model="gemma-4-E4B", model_provider="openai", api_key="dummy", base_url="http://localhost:8080/v1")
gemma_agent = create_agent(model=gemma_model)
```

```python
gemini_model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
gemini_agent = create_agent(model=gemini_model)
```

```python
from langchain.messages import HumanMessage

gemma_response = gemma_agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]}
)

gemini_response = gemini_agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]}
)
```

```python
from pprint import pprint

pprint(gemma_response)
pprint(gemini_response)
```

```python
print(gemma_response['messages'][-1].content)
print(gemini_response['messages'][-1].text)
```

```python
from langchain.messages import AIMessage

gemma_response = gemma_agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)


gemini_response = gemini_agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)
```

```python
pprint(gemma_response)
pprint(gemini_response)
```

## Streaming Output

```python
for token, metadata in gemma_agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
